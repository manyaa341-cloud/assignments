```python
# Q1: What is the average fare by service type?

df.groupby("service_type")["fare"].mean().sort_values(ascending=False)
```

InDrive has the highest average fare at 573.51, followed by Careem (544.98), Rickshaw (486.64), and Uber (486.13).

```python
# Q2: Which city has the most trips?

df.groupby("city")["trip_id"].count().sort_values(ascending=False)
```

Islamabad has the most trips with 29 trips, followed by Lahore (26), Karachi (23), and Rawalpindi (22)

```python
# Q3: What is the average distance per trip?
df["distance_km"].mean()
```

The average distance per trip is 9.61 km

```python
# Q4: Compare trip ratings by gender

df.groupby("gender")["trip_rating"].mean()
```

Males give a higher average trip rating (3.33) than females (2.82)

```python
# Q5: Make a bar chart of service type vs. average fare

avg_fare = df.groupby("service_type")["fare"].mean()
plt.bar(avg_fare.index, avg_fare.values, color='green', width=0.5)
plt.xlabel('Service Type')
plt.ylabel('Average Fare')
plt.title('Average Fare Bar Chart')
plt.show()
```
